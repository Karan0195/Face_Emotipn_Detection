```python
from keras.utils import to_categorical
from keras.preprocessing.image import load_img
from keras.models import Sequential
from keras.layers import Dense, Conv2D, Dropout, Flatten, MaxPooling2D
import os
import pandas as pd
import numpy as np
```

```python
TRAIN_DIR = "images/images/train"
TEST_DIR = "images/images/test"
```

```python
def createdataframe(dir):
    image_paths = []
    labels = []
    for label in os.listdir(dir):
        for imagename in os.listdir(os.path.join(dir, label)):
            image_path = os.path.join(dir, label, imagename)
            image_paths.append(image_path)
            labels.append(label)
        print(label, "complete")
    return image_paths, labels
```

```python
train = pd.DataFrame()
train['image'], train['label'] = createdataframe(TRAIN_DIR)

```

```python
print(train)
```

```python
test = pd.DataFrame()
test['image'], test['label'] = createdataframe(TEST_DIR)
```

```python
print(test)
```

```python
from tqdm.notebook import tqdm
```

```python
from tensorflow.keras.preprocessing.image import img_to_array
```

```python
def extract_features(images):
    features = []
    for image in tqdm(images):
        # Use color_mode instead of grayscale
        img = load_img(image, color_mode='grayscale', target_size=(48, 48))
        img = img_to_array(img)
        features.append(img)
    features = np.array(features)
    return features.reshape(len(features), 48, 48, 1)

    
```

```python
train_features = extract_features(train['image'])
```

```python
test_fetures = extract_features(test['image'])
```

```python
x_train = train_features/255.0
x_test = test_fetures/255.0
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
le.fit(train['label'])
```

```python
y_train = le.transform(train['label'])

y_test =le.transform(test['label'])
```

```python
y_train = to_categorical(y_train,num_classes = 7)
y_test = to_categorical(y_test,num_classes = 7)
```

```python
model = Sequential()

#convolutional Layers

model.add(Conv2D(128, kernel_size=(3,3), activation="relu", input_shape=(48,48,1)))
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.4))

model.add(Conv2D(256, kernel_size=(3,3), activation="relu")) 
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.4))

model.add(Conv2D(512, kernel_size=(3,3), activation  = "relu")) 
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.4))
    
model.add(Conv2D(512, kernel_size=(3,3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.4))

model.add(Flatten())

#fully connected Layers

model.add(Dense(512, activation="relu"))
model.add(Dropout(0.4))
model.add(Dense(256, activation='relu'))
model.add(Dropout(0.3))

#output Layer

model.add(Dense (7, activation='softmax'))
```

```python
model.compile(optimizer = 'adam', loss = 'categorical_crossentropy', metrics = ['accuracy'])
              
```

```python
model.fit(x= x_train , y= y_train, batch_size = 128, epochs = 100 , validation_data = (x_test,y_test))
```

```python
# Save model architecture to JSON
model_json = model.to_json()
with open("emotiondetector.json", "w") as json_file:
    json_file.write(model_json)   # typo fixed: erite → write

# Save model weights
model.save("emotiondetector.h5")  # typo fixed: hs → h5
from keras.models import model_from_json

# Load JSON file
with open("emotiondetector.json", "r") as json_file:
    loaded_model_json = json_file.read()

# Create model from JSON
loaded_model = model_from_json(loaded_model_json)

# Load weights into the model
loaded_model.load_weights("emotiondetector.h5")

print("Model loaded successfully!")
```

```python
label = ['angry','disgust','fear','happy','neutral','sad','surprise']
```

```python
# def ef(image):
#     img = load_img(image,grayscale = True )
#     feature = np.array(img)
#     feature.reshape(1,48,48,1)
#     return feature/255.0

def ef(image):
    # Load image in grayscale mode and resize to 48x48
    img = load_img(image, color_mode="grayscale", target_size=(48, 48))
    
    # Convert to numpy array
    feature = img_to_array(img)
    
    # Normalize
    feature = feature / 255.0
    
    # Reshape to (1, 48, 48, 1) for model input
    feature = np.reshape(feature, (1, 48, 48, 1))
    
    return feature
```
